```python
# Import libraries
import pandas as pd
import numpy as np
import plotly.express as px
```

# Data Loading

```python
from dotenv import load_dotenv

load_dotenv() # Load environment variables from .env dile
```

```python
import mysql.connector
import os

os.getenv('MYSQL_USER')
```

```python
conn = mysql.connector.connect(
    host = os.getenv('MYSQL_HOST'),
    user = os.getenv('MYSQL_USER'),
    password = os.getenv('MYSQL_PASSWORD'),
    database = os.getenv('MYSQL_DATABASE')
)
```

- Load customers

```python
query = '''
SELECT * FROM customers;
'''
customers = pd.read_sql(sql = query, con = conn)
customers.head(2)
```

- Load orders

```python
query = '''
SELECT * FROM orders;
'''
orders = pd.read_sql(sql = query, con = conn)
orders.head(2)
```

- Load products

```python
query = '''
SELECT * FROM products;
'''
products = pd.read_sql(sql = query, con = conn)
products.head(2)
```

- Load sales

```python
query = '''
SELECT * FROM sales;
'''
sales = pd.read_sql(sql = query, con = conn)
sales.head(2)
```

- Merge 'customers' with 'orders'

```python
df = pd.merge(left = customers, right = orders, how = 'inner', on = 'customer_id')
df.head(2)
```

- Merge with 'sales'

```python
df = pd.merge(left = df, right = sales, how = 'inner', on = 'order_id')
df.head(2)
```

- Merge with 'products'

```python
df = pd.merge(left = df, right = products, how = 'inner', on = 'product_id')
df.head(2)
```

# Data Cleaning

```python
# Drop unnecessary columns
df.drop(['country', 'home_address', 'zip_code', 'description'], axis = 1, inplace = True)
df.duplicated().sum()
```

```python
df.head()
```

```python
# Convert order_date and delivery_date to datetime
df['order_date'] = pd.to_datetime(df['order_date'])
df['delivery_date'] = pd.to_datetime(df['delivery_date'])
```

- In depth check for numerical columns

```python
num_cols = df.select_dtypes(include = 'number').columns.drop(['customer_id', 'order_id', 'product_id', 'sales_id'])
num_cols
```

```python
for col in num_cols:
    px.histogram(df, x = col, title = f'Histogram of {col}').show()
```

# Feature Engineering

```python
df['order_date'].describe()
```

```python
# Create month, day and weekday columns from order_date
# df['order_year'] = df['order_date'].dt.year won't do it because all the orders are in the same year (2023), so we will only create month and day columns
df['order_month_name'] = df['order_date'].dt.month_name()
df['order_day'] = df['order_date'].dt.day
df['order_weekday_name'] = df['order_date'].dt.day_name()
```

```python
df['delivery_days'] = (df['delivery_date'] - df['order_date']).dt.days
```

# Data Analysis

```python
# How many customers per state?
df.groupby('state')['customer_id'].nunique()
```

```python
from cProfile import label


px.bar(df.groupby('state')['customer_id'].nunique().sort_values(ascending=True).reset_index(), y = 'state', x = 'customer_id', title = 'Number of Customers per State', text_auto=True,
       labels={'customer_id': 'Number of Customers'}
       ).show()
```

```python
# Sales trend over time
sorted_df = df.sort_values(by = 'order_date')
plot_df = sorted_df.groupby('order_date')['total_price'].sum().reset_index()
plot_df
```

```python
px.line(plot_df, x = 'order_date', y = 'total_price', title = 'Sales Trend Over Time', labels={'total_price': 'Total Sales ($)', 'order_date': 'Order Date'}).show()
```

```python
df['month'] = df['order_date'].dt.month
df
```

```python
# Sales trend per month with values on each point
sorted_df = df.sort_values(by = 'month', ascending = True)
plot_df = sorted_df.groupby('month')['total_price'].sum().reset_index()
px.line(plot_df, x = 'month', y = 'total_price',
        title = 'Sales Trend Over Time', labels={'total_price': 'Total Sales ($)', 'month': 'Order Month'}, text = 'total_price').update_traces(textposition='top center').show()
```
