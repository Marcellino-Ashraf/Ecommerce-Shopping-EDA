# Data Cleaning

```python
# Drop column --> High Nan Percentage
df.drop('Cabin', axis= 1, inplace= True)
```

```python
# Drop Missing values raws in column --> Very low Nan Percentage
df.dropna(subset= 'Embarked', inplace= True, ignore_index= True)
```

```python
df.isna().mean().round(4) * 100
```

# Missing data handling

```python
px.histogram(data_frame= df, x = 'Age')
```

```python
# Age (Discrete) --> Fill with Median

age_median = df['Age'].median()
age_median
```

```python
df['Age'] = df['Age'].fillna(age_median)
df['Age'] 
```

```python
px.histogram(data_frame= df, x = 'Age')
```

```python
df.head()
```

### Another way to fill with median

```python
# Fill Age based on Gender group
df.groupby('Sex')['Age'].median()
```

```python
# [1, 2, 3, Nan, 5, 6, 7] --> Male
# [1, 2, 3, 4, 5, Nan, 7] --> Female
```

```python
df['Age'] = df.groupby('Sex')['Age'].transform(lambda x : x.fillna(x.median()))
df['Age']
```

```python
px.histogram(data_frame= df, x = 'Age')
```

## Using KNN imputer

```python
num_cols = df.select_dtypes(include= 'number').columns
num_cols
```
